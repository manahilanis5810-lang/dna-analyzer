```python
dna = "ATCCGATTGGGGAATCGTAATCG"
print("A:", dna.count("A"))
print("T", dna.count("T"))
print("G", dna.count("G"))
print("C", dna.count("C"))
```

```python
dna = "ATCCGATTGGGGAATCGTAATCG"
g = dna.count("G")
c = dna.count("C")
gc_content = (g+c) / len(dna) * 100
print ("GC content", gc_content, "%")
```

```python
dna = "ATGCGCTAGCATGCGCTAGC"
complement = ""
for base in dna:
    if base == "A":
        complement += "T"
    elif base == "T":
        complement += "A"
    elif base == "G":
        complement += "C"
    elif base == "C":
        complement += "G"
print(dna)
print(complement)
```

```python
dna = "ATGCGCTAGCATGCGCTAGC"
for i in range(0, len(dna) - 2, 3):
  codon = dna [i:i+3]
  print(codon)
```

```python
codon_table = {
    "ATG": "Methionine (start)",
    "CGC": "Arginine",
    "TAG": "Stop",
    "AGC": "Serine",
    "GCG": "Alanine",
    "TGC": "Cysteine",
}

dna = "ATGCGCTAGCATGCGCTAGC"
for i in range(0, len(dna) - 2, 3):
    codon = dna[i:i+3]
    amino_acid = codon_table.get(codon, "Unknown")
    print(codon, "->", amino_acid)
```
